# Face Recognition Setup and loyalty visits logging
This notebook sets up the **OpenCV LBPH Face Recognizer** to recognize customers (Alice, Bob, Charlie) and logs visits for customer loyalty analytics.

```python
import os
import cv2
import joblib
import numpy as np
```

## 1. Load the Customer Face Database and XML Model

```python
db_path = '../app/models/face_db.pkl'
xml_path = '../app/models/face_recognizer.xml'

if os.path.exists(db_path) and os.path.exists(xml_path):
    customer_db = joblib.load(db_path)
    recognizer = cv2.face.LBPHFaceRecognizer_create()
    recognizer.read(xml_path)
    print("Loaded customer database and LBPH model successfully!")
    print(customer_db)
else:
    print("Models not found. Run training script first.")
```

## 2. Face Verification Pipeline (Detect -> Recognize -> Log Visit)

```python
# Let's mock a face input image for Bob (ID 2)
test_img = np.zeros((100, 100), dtype=np.uint8)
cv2.circle(test_img, (50, 50), 40, 200, -1)
cv2.rectangle(test_img, (30, 35), (40, 45), 50, -1) # left eye
cv2.rectangle(test_img, (60, 35), (70, 45), 50, -1) # right eye
cv2.line(test_img, (35, 65), (65, 65), 50, 3) # Bob's mouth

# Predict using the LBPH recognizer
label, confidence = recognizer.predict(test_img)
print(f"Predicted customer ID: {label} (Confidence: {confidence:.2f})")

if label in customer_db and confidence < 50:
    customer = customer_db[label]
    print(f"Welcome back, {customer['name']}!")
    print(f"Loyalty Points: {customer['loyalty_points']}")
else:
    print("Unknown customer detected. Would you like to register?")
```
